```python
import os
import tabula
import pandas as pd

# Java path
os.environ["JAVA_HOME"] = r"C:\Program Files\Java\jdk-25.0.2"
os.environ["PATH"] = os.environ["JAVA_HOME"] + r"\bin;" + os.environ["PATH"]

pdf_path = r"D:\Documents\GitHub\MISC\ConvertMerge\PSTA_5_Fulltext__Final.pdf"
output_excel = r"D:\Documents\GitHub\MISC\ConvertMerge\PSTA5_tables.xlsx"

# Extract tables from ALL pages
tables = tabula.read_pdf(
    pdf_path,
    pages="104",
    multiple_tables=True,
    lattice=True, stream=True
)

print(f"Total tables detected: {len(tables)}")

# Save to Excel with multiple sheets
with pd.ExcelWriter(output_excel, engine="openpyxl") as writer:
    for i, table in enumerate(tables):
        sheet_name = f"Table_{i+1}"
        table.to_excel(writer, sheet_name=sheet_name, index=False)

print("All tables saved to Excel!")
```
